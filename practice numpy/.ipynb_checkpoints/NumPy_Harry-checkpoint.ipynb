```python
# --------------------------------
range = np.arange(15)
range 
# this gives the numpy array of 0 to (n-1) array 
```

```python
# -----------------------
lspace = np.linspace(1,10,5)
lspace 
# In this 1- first element, 10 - last element, 5 - is the total content present.
```

```python
empty = np.empty((2,5))
empty 
# it does not always fill with 0s, just allocates memory with garbage value.
# used when you want to fill the array with actual data.
```

```python
emp = np.empty_like(lspace)
emp
# this is more efficient and less space consuming
```

```python
# -------------------
identity_matrix = np.identity(5)
identity_matrix
```

```python
a = np.array([1,2,3,4,5,6])
# a = np.arange(7)
# the above both are same way to represent


# at the end both tells the dimentions only
a.shape # tell the length 5
a.ndim # tell the Number of dimensions: 1D
```

```python
# if i want suppose want to convert the 1D into 2 by 3 of a NOTE; it will be coverted in 2 by3 only since it has 6 elements
y= a.reshape((2,3)) #2x3 = 6, 3x2 =6 
y
```

```python
# as you can see the digits are arranged rows wise, what if i want to arrange it in col wise then
x= a.reshape((2,3),order = "F")
x
```

```python
# --------------------------
# using of ravel to make it as it was befor into !D
x.ravel()
```

```python
# ------------------
# NumPy Axis
z = [[1,2,3],[4,5,6],[7,1,0]]
array = np.array(z)
array 
```

```python
array.sum(axis=1)
# this is gives the sum of the x axis elements
```

```python
# Methods and attributes 
array.T
# this is gives the transpose of the 'array'
```
