```python
# Two Sum 
nums = [2, 7, 11, 15]
target = 9

n = len(nums)
for i in range (n):
    for j in range(i+1,n):
        if nums[i] + nums[j] == target:
            print([i,j])

# this is solved using the brute force.
# TC - O(n2)
# SP - O(1)

# alternative method 
# hashmap method using dict
# in this we use the first 2 elements and find there comiliment and enumerate.
# TC - O(n)
# SC - O(n)
```

```python
# Equiberlium - pivotal index 
nums = [1,7,3,6,5,6]

for i in range(len(nums)):
    left_sum = sum(nums[:i])
    right_sum = sum(nums[i+1:])

    if left_sum == right_sum:
        print(i)

# ✅ Time Complexity = O(n²) - if nested loops are made. 
# ✅ Space Complexity = O(n) - if any dict set list is created
```

```python
# ✅ Time Complexity (TC)

# 1 loop                 → O(n)

# 2 nested loops         → O(n²)

# 3 nested loops         → O(n³)

# Sorting                → O(n log n)

# HashMap lookup         → O(1)

# ✅ Space Complexity (SC)
# Only variables → O(1)
# list, dict, set, stack - O(n)
```

```python
# Find Duplicate Element in an Array
arr = [1, 2, 3, 4, 2]

for i in range(len(arr)):
    for j in range(i+1,len(arr)):
        if arr[i] == arr[j]:
            print(arr[i])

# TC - O(n2)
# SC - O(n) 
```

```python
# Most Frequent Element in an Array.
arr = [1, 2, 2, 3, 4, 2, 5]

max_count = 0
most_frequent = arr[0]

for i in arr:
    count = arr.count(i)

    if count>max_count:
        max_count = count
        most_frequent = i
    
    print("the most repeated one is :", most_frequent)



# TC - O(n2) 
# No extra list, set, or dictionary is created. so SC - O(1)
```

```python
# Unique Element in Array
arr = [1, 2, 2, 3, 3, 4]

for i in range(len(arr)):
    count = 0 
    for j in range(len(arr)):

        if arr[i] == arr[j]:
            count += 1
        
    if count == 1:
        print(arr[i])

# TC - O(n2) 
# No extra list, set, or dictionary is created. so SC - O(1)  
```

```python
# alternative
arr = [1, 2, 2, 3, 3, 4]

for i in arr:
    if arr.count(i) == 1:
        print(i)
```
