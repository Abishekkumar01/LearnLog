# if else and elif 

```python
# Usage of if 
age = 20

if age>= 18:
    print("you can vote")
```

```python
# Usage of if and else 
age = 14

if age>= 18:
    print("you can vote")
else:
    print("you are a minor")
```

```python
# Usage of if, else, elif 
age = int(input("enter you age"))

if age<18:
    print("you are a child")
elif age<=30:
    print("you are a adult")
else:
    print("you are a senior citizen")
```

```python
#Nested conditional statement
# Write a Python program using nested if statements to check whether a user-entered number is positive, negative, or zero, and if positive, determine whether it is even or odd.

num = float(input("Enter the number"))

if num>0:
    print("It is a postive number.")
    if num%2==0:
        print("it is a even number.")
    else:
        print("it is a odd number.")
else:
    print("it is a zero or a negative number.")
```

```python
# to find the year is a lead year or not 

year = int(input("Enter the year:"))

if (year %4==0 and year %100 != 0 ) or (year%400 == 0):
    print("It is a lead year")
else:
    print("It is a not a leap year")
```

```python
# Same Question in nested form 
year = int(input("Enter the Year:"))

if year%4 == 0:
    if year%100 == 0:
        if year%400 == 0:
            print("leap year")
        else:
            print("Not a leap year")
    else:
        print("leap year")
else:
    print("Not a leap year")
```

```python
# Build an calculator using nested loop 
num1 = float(input("Enter the Num1:"))
num2 = float(input("Enter the Num2:"))
operations = input("Enter the value (+,-,*,/)")

if operations == '+':
    result = num1 + num2
elif operations == '-':
    result = num1 - num2
elif operations == '*':
    result = num1 * num2
elif operations == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error, not divisible by 0"
else:
    result = "Invalid operation"

print("Result:", result)

```
