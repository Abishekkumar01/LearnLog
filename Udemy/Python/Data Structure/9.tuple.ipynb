```python
# 7. Packing and Unpacking Tuples
# packing 
pack_now = 3.14,"abi",1,55
print(pack_now)
```

```python
# how to unpack - 2 methods 
a,b,c,d = pack_now
print(a)
print(b)
print(c)
print(d)
```

```python
# unpack using *
first,*middle,last = pack_now
print(first)
print(middle)
print(last)

# this gives btw all elements of first and last 
```

```python
# 8. Nested Tuples

# first nested list 
listt = [[1,2,3,4,5],[6,7,8,9],[3.14,5,"abhi",True]]
print(listt[0]) # i can access an eleement also
print(listt[0][:3]) # i can slice also 
```

```python
# nested list with tuple inside
value = [[1,2,3,4,5],[6,7,8,9],(3.14,5,"abhi",True)]
print(value[2])
print(value[2][:3])
# still it works the same 
```

```python
# nested tuple 
values = (( 1,2,3,4,5),(6,7,8,9),(3.14,5,"abhi",True))
print(values[0])
print(values[0][:3])
# still it works the same   
```

```python
# iterating over the nested tuple 
for sub_tuple in values:
    for items in sub_tuple:
        print(items, end = " ")
    print()
```
