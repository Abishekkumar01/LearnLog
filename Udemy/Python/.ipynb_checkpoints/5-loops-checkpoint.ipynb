# Loops

```python
range(5)
```

```python
# for loop
for i in range (10):
    print(i)
```

```python
for i in range(1,10):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
# coz of the -2 it is going from back 
for i in range(10,1,-2):
    print(i)
```

```python
str = "Abishek"

for i in str:
    print(i)
```

```python
# while loop 
count = 0

while count<5:
    print(count)
    count = count+1
```

```python
count = 0
while count%2 ==0:
    print(count)
    count=count+1
```

```python
# break statement
# break statement exits the loop prematuely
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
# continue statement
# it skips the current iteration and continues with next
# eg. this is gives odd number
for i in range(10):
    if i%2==0:
        continue
    print(i)


# if naturally if wanted to generate the odd numbers then
# for i in range(10):
#     if i%2!=0:
#         print(i)
```

```python
# pass - means do notthing for now (suppose you want to write the code later)
for i in range(1,6):
    if i==3:
        pass
    print(i)
```

```python
# nested loops
# eg. to print * in box 
for i in range(3):
    for j in range(4):
        print("*", end = " ")
    print()
```

```python
# now if to print triangle
for i in range(1,5):
    for j in range(i):
        print("*", end= " ")
    print()
```
